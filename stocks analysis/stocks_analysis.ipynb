# Stocks Analysis Based on data extracted from Investing.com

## Porcessing Single File

```python
import pandas as pd
from datetime import datetime
from sqlalchemy import create_engine
```

```python
file_path = r"D:\Investing.com\EGX30\23012025\EGX 30 (EGX30)_Price.csv"
```

```python
hostname = "localhost"
dbname = "stg_stocks_analysis"
uname = "root"
pwd = "production_server"
```

```python
df = pd.read_csv(file_path,)
```

```python
df=df.drop(columns=['Upside'])
```

```python
df['AS_OF_DATE'] = datetime.today().strftime('%Y-%m-%d')
```

```python
df
```

```python
engine = create_engine(f"mysql+pymysql://{uname}:{pwd}@{hostname}/{dbname}")
```

```python
df.to_sql("stocks_price", engine, if_exists="replace", index=False)
```
